### Los medicos {}

```python
# El directorio principal ahora es un diccionario vacío utilizando llaves
directorio_medicos = {}

# --- Funciones del Sistema ---

def registrar_medico():
    print("\n--- Entrada de Datos: Registro de Especialista ---")
    nombre = input("IA: Nombre del médico: ")

    # Creamos la ficha del médico usando llaves
    medico_nuevo = {
        "Especialidad": input("IA: Especialidad (general / pediatra / cirujano): ").lower(),
        "Experiencia": int(input("IA: Años de trayectoria: ")),
        "Pacientes": int(input("IA: Pacientes atendidos: "))
    }

    # Guardamos en el diccionario principal usando el nombre como clave
    directorio_medicos.update({nombre: medico_nuevo})
    print("SISTEMA: Médico guardado en el directorio.")

def mostrar_directorio_completo():
    # Verificamos si el diccionario está vacío
    if not directorio_medicos:
        print("\nIA AVISO: El directorio está vacío. Registre médicos primero.")
    else:
        print("\n" + "="*70)
        print("{:<15} | {:<15} | {:<10} | {:<10}".format("NOMBRE", "ESPECIALIDAD", "EXP (AÑOS)", "PACIENTES"))
        print("-" * 70)

        # Recorremos el diccionario (llave: nombre, valor: datos del médico)
        for nombre, datos in directorio_medicos.items():
            print("{:<15} | {:<15} | {:<10} | {:<10}".format(
                nombre,
                datos.get("Especialidad"),
                datos.get("Experiencia"),
                datos.get("Pacientes")
            ))
        print("="*70)

def consultar_ia():
    if not directorio_medicos:
        print("\nIA ERROR: No hay personal para analizar.")
    else:
        print("\n--- Motor de Lógica IA: Recomendación ---")
        sintoma = input("IA: Síntoma del paciente (fiebre / niños / fractura): ").lower()
        encontrado = False

        for nombre, datos in directorio_medicos.items():
            esp = datos.get("Especialidad")
            exp = datos.get("Experiencia")

            if sintoma == "fiebre" and esp == "general":
                print("IA RECOMIENDA: Al Dr(a). " + nombre + " (Especialista General).")
                encontrado = True
                break
            elif sintoma == "niños" and esp == "pediatra":
                print("IA RECOMIENDA: Al Dr(a). " + nombre + " (Ideal para infantes).")
                encontrado = True
                break
            elif sintoma == "fractura" and esp == "cirujano" and exp >= 10:
                print("IA RECOMIENDA: Al Dr(a). " + nombre + " por su alta experiencia.")
                encontrado = True
                break

        if not encontrado:
            print("IA NOTA: No hay un perfil exacto para este caso en el directorio actual.")

# --- Programa Principal (Menú Interactivo) ---

while True:
    print("\n--- SISTEMA MÉDICO IA (MENÚ PRINCIPAL) ---")
    print("1. Registrar nuevo médico")
    print("2. Mostrar tabla de TODOS los médicos")
    print("3. Consultar recomendación de la IA")
    print("4. Apagar sistema")

    opcion = input("Seleccione una opción (1-4): ")

    if opcion == "1":
        registrar_medico()
    elif opcion == "2":
        mostrar_directorio_completo()
    elif opcion == "3":
        consultar_ia()
    elif opcion == "4":
        print("IA: Cerrando base de datos. ¡Adiós!")
        break
    else:
        print("ERROR: Opción no válida.")

print("\nElaborado por Alexandro Aguilar NL = 1")
```
